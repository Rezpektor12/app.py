```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = (224, 224)
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    validation_split=0.2
)

train_generator = train_datagen.flow_from_directory(
    base_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    subset='training'
)

validation_generator = train_datagen.flow_from_directory(
    base_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    subset='validation'
)
```

```python
validation_generator = train_datagen.flow_from_directory(
    "/content/sample_data/Data_Train",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    subset='validation'
)
```

```python
from tensorflow.keras.applications import MobileNetV2 # Import MobileNetV2

base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(244, 244, 3)) # Use MobileNetV2

# Unfreeze some layers for fine-tuning
num_unfreeze_layers = 20  # You can adjust this number
for layer in base_model.layers[-num_unfreeze_layers:]:
    layer.trainable = True

# Keep the rest of the layers frozen
for layer in base_model.layers[:-num_unfreeze_layers]:
    layer.trainable = False
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

# Gunakan model ResNet50 sebagai base model (tanpa fully connected layer bawaan)
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))

# Bekukan layer dasar agar tidak ikut dilatih pada awal
for layer in base_model.layers:
    layer.trainable = False

# Buat layer tambahan di atas base model
x = base_model.output
x = GlobalAveragePooling2D()(x)  # Rata-rata dari peta fitur (lebih efisien dari Flatten)
x = Dense(512, activation='relu')(x)
x = Dropout(0.5)(x)
output_layer = Dense(1, activation='sigmoid')(x)  # Binary classification (Pisang vs Kelapa)

# Gabungkan base model + layer tambahan
model = Model(inputs=base_model.input, outputs=output_layer)

# Kompilasi model
model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Ringkasan model
model.summary()
```

```python
from tensorflow.keras.optimizers import Adam

# 1️⃣ Buka sebagian layer pada base_model agar bisa dilatih ulang
for layer in base_model.layers[-10:]:  # buka 10 layer terakhir (bisa diubah sesuai kebutuhan)
    layer.trainable = True

# 2️⃣ Recompile model dengan learning rate yang sangat kecil
model.compile(
    optimizer=Adam(learning_rate=0.00001),  # Learning rate kecil untuk fine-tuning
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 3️⃣ Lanjutkan training
history_finetune = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=5,
    verbose=1
)
```

```python
# Latih ulang model setelah fine-tuning
history_finetune = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20,        # Tambah jumlah epoch
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

# Ekstrak nilai accuracy dari riwayat
train_acc = history_finetune.history['accuracy']
val_acc = history_finetune.history['val_accuracy']

# Ekstrak nilai loss dari riwayat
train_loss = history_finetune.history['loss']
val_loss = history_finetune.history['val_loss']

# Buat range untuk jumlah epoch
epochs = range(1, len(train_acc) + 1)

# Plot Accuracy
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, train_acc, 'bo-', label='Training Accuracy')
plt.plot(epochs, val_acc, 'r^-', label='Validation Accuracy')
plt.title('Training & Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Plot Loss
plt.subplot(1, 2, 2)
plt.plot(epochs, train_loss, 'bo-', label='Training Loss')
plt.plot(epochs, val_loss, 'r^-', label='Validation Loss')
plt.title('Training & Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

# Tampilkan grafik
plt.show()
```

```python
# Evaluasi model pakai validation data sebagai test
test_loss, test_acc = model.evaluate(validation_generator, verbose=1)

print(f"\n✅ Validation Accuracy (as Test) : {test_acc:.4f}")
print(f"✅ Validation Loss               : {test_loss:.4f}")
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# 🔹 Ambil semua gambar & label dari test_generator
# (Pastikan test_generator sudah didefinisikan sebelumnya)
test_images_all, test_labels_all = [], []
for _ in range(len(test_generator)):
    images, labels = next(test_generator)
    test_images_all.extend(images)
    test_labels_all.extend(labels)

test_images_all = np.array(test_images_all)
test_labels_all = np.array(test_labels_all)

# 🔹 Gunakan model yang sudah dimuat untuk prediksi seluruh data uji
predictions_all = loaded_model.predict(test_images_all, verbose=1)
predicted_classes_all = (predictions_all > 0.5).astype(int).flatten()

# 🔹 Tentukan jumlah gambar yang ingin ditampilkan
numbers_to_display = min(25, len(test_images_all))
num_cells = math.ceil(math.sqrt(numbers_to_display))

plt.figure(figsize=(15, 15))

for plot_index in range(numbers_to_display):
    plt.subplot(num_cells, num_cells, plot_index + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)

    # 🔹 Tentukan label prediksi & label sebenarnya
    predicted_label_name = 'Kelapa 🥥' if predicted_classes_all[plot_index] == 1 else 'Pisang 🍌'
    true_label_name = 'Kelapa 🥥' if test_labels_all[plot_index] == 1 else 'Pisang 🍌'

    # 🔹 Warna hijau kalau benar, merah kalau salah
    color_map = 'Greens' if predicted_classes_all[plot_index] == test_labels_all[plot_index] else 'Reds'

    plt.imshow(test_images_all[plot_index], cmap=color_map)
    plt.xlabel(
        f"Pred: {predicted_label_name}\nTrue: {true_label_name}",
        color=('green' if predicted_classes_all[plot_index] == test_labels_all[plot_index] else 'red'),
        fontsize=9
    )

plt.subplots_adjust(hspace=0.6, wspace=0.4)
plt.suptitle("🔍 Hasil Prediksi Model Pisang vs Kelapa", fontsize=16)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sn
import tensorflow as tf
import numpy as np

# Pastikan test_labels_all dan predicted_classes_all sudah dalam bentuk numpy array
test_labels_all = np.array(test_labels_all).astype(int)
predicted_classes_all = np.array(predicted_classes_all).astype(int)

# 🔹 Hitung confusion matrix
confusion_matrix = tf.math.confusion_matrix(test_labels_all, predicted_classes_all)

# 🔹 Tampilkan confusion matrix dalam bentuk heatmap
plt.figure(figsize=(8, 6))
sn.heatmap(
    confusion_matrix,
    annot=True,
    cmap='Blues',
    fmt='d',
    linewidths=.5,
    xticklabels=['Pisang 🍌', 'Kelapa 🥥'],  # Label prediksi
    yticklabels=['Pisang 🍌', 'Kelapa 🥥']   # Label sebenarnya
)
plt.xlabel('Prediksi')
plt.ylabel('Label Sebenarnya')
plt.title('📊 Confusion Matrix - Klasifikasi Pisang vs Kelapa')
plt.show()
```
